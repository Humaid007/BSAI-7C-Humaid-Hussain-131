## Case-by-case findings

Read `PART_A_REPORT.md` for all nine detailed analyses. The next cell reloads the CSVs and verifies the main self-check numbers from the assignment.

```python
helmet=pd.read_csv(DATA/'case01_helmet_rickshaw.csv')
helmet_totals=helmet.groupby('helmet_grade')[['head_injuries','trips']].sum()
helmet_totals['injuries_per_1000']=1000*helmet_totals.head_injuries/helmet_totals.trips
chai=pd.read_csv(DATA/'case02_chai_quizzes.csv')
quartet=pd.read_csv(DATA/'case04_quartet_depts.csv')
shapes=pd.read_csv(DATA/'case05_same_stats_shapes.csv')
sensor=pd.read_csv(DATA/'case06_sensor_matrix.csv')
vitals=pd.read_csv(DATA/'vitals_10k.csv')
print('Helmet rates per 1,000 trips:\n',helmet_totals['injuries_per_1000'])
print('Chai/quiz Pearson r:',chai.cups_doodh_patti_sold.corr(chai.quiz_failures))
print('Quartet summaries:\n',quartet.groupby('department').agg(n=('study_hours_per_week','size'),mean_hours=('study_hours_per_week','mean'),mean_score=('final_score_out_of_20','mean')))
print('Shape names:',shapes.shape_name.unique())
print('Sensor correlation matrix shape:',sensor.select_dtypes(include=np.number).shape)
print('Observed max SpO2:',vitals.spo2.max())
```
