# Week 3 — Part B: Lie Factor

This notebook reproduces the calculations from the provided `case07_lie_factor.csv` and `vizlib.py`. Keep both files in the same folder as this notebook.

```python
from pathlib import Path
import pandas as pd
from vizlib import lie_factor, lie_verdict

DATA = Path("case07_lie_factor.csv")
df = pd.read_csv(DATA)
df.head()
```

## Calculate data effect, graphic effect, and lie factor

`ink_dimension=2` is used for symbols where the encoding is area (both width and height, or bubble area).

```python
rows = []
for _, r in df.iterrows():
    lf, data_effect, graphic_effect = lie_factor(
        r.value_start, r.value_end, r.ink_start, r.ink_end,
        ink_dimension=int(r.ink_dimension)
    )
    rows.append({
        "chart_id": r.chart_id,
        "subject": r.subject,
        "data_effect": data_effect,
        "graphic_effect": graphic_effect,
        "lie_factor": lf,
        "verdict": lie_verdict(lf),
    })
results = pd.DataFrame(rows)
results
```

```python
results.to_csv("lie_factor_results.csv", index=False)
print("Saved lie_factor_results.csv")
```
